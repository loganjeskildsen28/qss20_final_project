```python
## Read all relevant datasets

## Basic info
basic_info_df = pd.read_csv('../data/basic_info.csv')

## Wikipedia Information
article_views_df = pd.read_csv('../data/article_views.csv')
country_views_df = pd.read_csv('../data/country_article_views.csv')
football_views_df = pd.read_csv('../data/football_views.csv')

## Google Trends information
regional_search_interest_df = pd.read_csv('../data/regional_search_interest.csv')
```

```python
## First, isolate for culture related articles (culture of {country}, history of {country}, society of {country})

culture_views_df = article_views_df[(article_views_df['subcategory_title_fixed'].str[:10] == 'Culture of') | 
    (article_views_df['subcategory_title_fixed'].str[:10] == 'History of') |
    (article_views_df['subcategory_title_fixed'].str[:10] == 'Society of')].copy().reset_index(drop=True)

culture_views_df
```

```python
## Now subset for the first ten days 5-17-2026 - 5-26-2026, this will act as a baseline for calculating means, standard deviation, etc.

culture_baseline_df = culture_views_df[(culture_views_df['date'] >= 20260517) & (culture_views_df['date'] <= 20260526)].copy().reset_index(drop=True)

culture_baseline_df

## Do the same with the country and football datasets to calculate respective baselines.

football_baseline_df = football_views_df[(football_views_df['date'] >= 20260517) & (football_views_df['date'] <= 20260526)].copy().reset_index(drop=True)

football_baseline_df

country_baseline_df = country_views_df[(country_views_df['date'] >= 20260517) & (country_views_df['date'] <= 20260526)].copy().reset_index(drop=True)

country_baseline_df

## Now calculate baseline means, standard deviations for each dataframe

culture_baseline_stats = (
    culture_baseline_df.groupby(['country', 'article_title'])['views']
    .agg(mean_views='mean', std_views='std')
    .reset_index()
    .copy()
                         )

culture_baseline_stats

football_baseline_stats = (
    football_baseline_df.groupby(['country', 'article_title'])['views']
    .agg(mean_views='mean', std_views='std')
    .reset_index()
    .copy()
                         )

football_baseline_stats

country_baseline_stats = (
    country_baseline_df.groupby(['country', 'article_title'])['views']
    .agg(mean_views='mean', std_views='std')
    .reset_index()
    .copy()
                         )

country_baseline_stats

culture_baseline_df = culture_baseline_df.drop_duplicates()

```

```python
## Left join basic_info with all dataframes, eliminate data points before 5-26-2026 as they are used in baseline calculations already

culture_filtered_df = culture_views_df[culture_views_df['date'] >= 20250527]
football_filtered_df = football_views_df[football_views_df['date'] >= 20250527]
country_filtered_df = country_views_df[country_views_df['date'] >= 20250527]

## Perform first left join
culture_total_df = pd.merge(culture_filtered_df, basic_info_df, how = 'left', on="country")
football_total_df = pd.merge(football_filtered_df, basic_info_df, how = 'left', on="country")
country_total_df = pd.merge(country_filtered_df, basic_info_df, how = 'left', on="country")

## Now perform second left join with all means, standard deviations for each article viewcount.
culture_final_df = pd.merge(culture_total_df, culture_baseline_df, how = 'left', on=["country", "article_title"])
football_final_df = pd.merge(culture_total_df, football_baseline_df, how = 'left', on=["country", "article_title"])
country_final_df = pd.merge(culture_total_df, country_baseline_df, how = 'left', on=["country", "article_title"])

culture_final_df
football_final_df
country_final_df
```
